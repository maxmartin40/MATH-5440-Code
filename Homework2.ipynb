```python
import numpy as np 
import matplotlib.pyplot as plt
```

### Problem 8 Definition and Solution
The solution of the diffusion equation we want to plot is defined as:
$$
u(x,t) = \sum_{n \text{ odd}} \frac{32}{n^3 \pi^3} \sin(n\pi x) e^{-n^2\pi^2 t}
$$

```python
def u(x,t):
    sol = 0
    for n in range(1,12,2):
        sol += 32/((n**3)*(np.pi**3)) * np.sin(n*np.pi*x)*np.e**(-1*n**2*np.pi**2*t)
    return sol
```

We will now make a line plot as given in the assignment.

```python
x = np.linspace(0,1,100)

plt.figure(figsize=(7, 4))

for t in [0, 0.01, 0.05, 0.2]:
    plt.plot(x, u(x, t), label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Diffusion Equation Solution at Different Times")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Now we will make a surface plot as directed in the assignment.

```python
x = np.linspace(0,1,100)
t = np.linspace(0,0.2,100)

X, T = np.meshgrid(x,t)

U = u(X,T)

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection='3d')

surface = ax.plot_surface(X, T, U, cmap="nipy_spectral", edgecolor="none")

ax.set_xlabel("x")
ax.set_ylabel("t")
ax.set_zlabel("u(x,t)")
ax.set_title("Surface Plot of the Wave Equation Solution")

fig.colorbar(surface, ax=ax, shrink=0.6, label="u(x,t)")

plt.tight_layout()
plt.show()
```
