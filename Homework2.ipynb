### Problem 8 Definition and Solution
The solution of the diffusion equation we want to plot is defined as:
$$
u(x,t) = \sum_{n \text{ odd}} \frac{32}{n^3 \pi^3} \sin(n\pi x) e^{-n^2\pi^2 t}
$$

```python
def u(x,t):
    total = 0
    sol = 0
    for n in range(1,11,2):
        sol += 32/((n**3)*(np.pi**3)) * np.sin(n*np.pi*x)*np.e**(-1*n**2*np.pi**2*t)
    return sol
```

We will now make a line plot as given in the assignment.

```python
x = np.linspace(-5,5,100)

plt.figure(figsize=(7, 4))

for t in [0, 0.01, 0.05, 0.2]:
    plt.plot(x, u(x, t), label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("Solution at Different Times")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
