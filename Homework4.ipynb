```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
```

Create contour plot

```python
R = 2
N = 400
x = np.linspace(-R, R, N)
y = np.linspace(-R, R, N)
X, Y = np.meshgrid(x, y)

U = 1 + 1.5 * X * Y

disk_mask = (X**2 + Y**2) <= R**2
U_disk = np.where(disk_mask, U, np.nan)

fig_contour, ax_c = plt.subplots(figsize=(6, 6))

contour_filled = ax_c.contourf(X, Y, U_disk, levels=50, cmap='viridis')
fig_contour.colorbar(contour_filled, ax=ax_c, label='u(x, y)')

contour_lines = ax_c.contour(X, Y, U_disk, levels=10, colors='black', linewidths=0.5)
ax_c.clabel(contour_lines, inline=True, fontsize=8, fmt='%.1f')

boundary = Circle((0, 0), R, color='black', fill=False, linewidth=2, linestyle='--')
ax_c.add_patch(boundary)

ax_c.set_title('Contour Plot of $u(x, y) = 1 + \\frac{3}{2}xy$ ($r \\le 2$)', fontsize=12)
ax_c.set_xlabel('X')
ax_c.set_ylabel('Y')
ax_c.set_aspect('equal')
ax_c.grid(True, linestyle=':', alpha=0.5)
```
