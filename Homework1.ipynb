```python
import numpy as np 
import matplotlib.pyplot as plt
```

Homework Problem 4(a): Define the solution to Problem 3, calculated by hand using d'Alembert's formula.

```python
def u(x,t):
    sol = 0.5*np.log((1+(x+t)**2)*(1+(x-t)**2))+4*t+x*t
    return sol
```

```python
x = np.linspace(-5,5,100)

plt.figure(figsize=(7, 4))

for t in [0, 0.5, 1]:
    plt.plot(x, u(x, t), label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("4(a) Wave Equation Solution at Different Times")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
