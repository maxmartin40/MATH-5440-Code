Import packages

```python
import numpy as np 
import matplotlib.pyplot as plt
```

Homework Problem 4(a): Define the solution to Problem 3, calculated by hand using d'Alembert's formula.

```python
def u(x,t):
    sol = 0.5*np.log((1+(x+t)**2)*(1+(x-t)**2))+4*t+x*t
    return sol
```

```python
x = np.linspace(-5,5,100)

plt.figure(figsize=(7, 4))

for t in [0, 0.5, 1]:
    plt.plot(x, u(x, t), label=f"t = {t}")

plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.title("4(a) Wave Equation Solution at Different Times")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Homework Problem 4(b): We now need to create a surface plot for the solutions.

```python
x = np.linspace(-5,5,100)
t = np.linspace(0,1,100)

X, T = np.meshgrid(x,t)

U = u(X,T)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

surface = ax.plot_surface(X, T, U, cmap="viridis", edgecolor="none")

ax.set_xlabel("x")
ax.set_ylabel("t")
ax.set_zlabel("u(x,t)")
ax.set_title("Surface Plot of the Wave Equation Solution")

fig.colorbar(surface, ax=ax, shrink=0.6, label="u(x,t)")

plt.tight_layout()
plt.show()
```
