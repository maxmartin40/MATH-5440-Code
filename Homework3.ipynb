Consider the function $\phi=x$ on $(0,\ell)$. On the same axis, plot the sum of the first three (nonzero) terms of the Fourier sine and cosine series.

```python
import numpy as np
import matplotlib.pyplot as plt
```

The sine and cosine series were taken from derivations in the textbook (Strauss).

```python
def sine_series(x,l,n):
    sine_sum = 0
    for i in range(0,n):
        phi = (2*l/np.pi)*(np.sin(np.pi*x/l)-0.5*np.sin(2*np.pi*x/l)+(1/3)*np.sin(3*np.pi*x/l))
        sine_sum += phi
    return sine_sum

def cosine_series(x,l,m):
    cosine_sum = 0
    for i in range(0,2*m):
        if m%2 == 0:
            cosine_sum += 0
        if m%2 == 1:
            phi = (-4*l)/(m**2 * np.pi**2)
            cosine_sum += phi
    return cosine_sum
```
