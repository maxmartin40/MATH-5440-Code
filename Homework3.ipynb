## Problem 6

Consider the function $\phi=x$ on $(0,\ell)$. On the same axis, plot the sum of the first three (nonzero) terms of the Fourier sine and cosine series.

```python
import numpy as np
import matplotlib.pyplot as plt
```

The sine and cosine series were taken from derivations in the textbook (Strauss).

```python
def sine_series(x,l,n):
    sine_sum = np.zeros_like(x,dtype=float)
    for i in range(1,n+1):
        phi = (2*l/np.pi)*(np.sin(np.pi*x/l)-0.5*np.sin(2*np.pi*x/l)+(1/3)*np.sin(3*np.pi*x/l))
        sine_sum += phi
    return sine_sum

def cosine_series(x,l,m):
    cosine_sum = (l/2) * np.zeros_like(x,dtype=float)
    for i in range(1,m+1):
        #phi = (l/2) - 4*l/(np.pi**2) * (np.cos(np.pi*x/l)+(1/9)*np.cos(3*np.pi(x)/l)+(1/25)*np.cos(5*np.pi*x/l))
        phi = (2 * l / (m * np.pi)**2) * (((-1)**m - 1) * np.cos(m * np.pi * x / l))
        cosine_sum += phi
    return cosine_sum
```

```python
L = 2*np.pi

x = np.linspace(0,L,1000)

plt.plot(x, sine_series(x, L, 3), label="Sine series")
plt.plot(x, cosine_series(x, L, 3), label="Cosine series")
plt.plot(x, x, "k--", label="f(x) = x")

plt.xlabel("x")
plt.ylabel("y")
plt.title("First 3 Fourier Series Terms for f(x) = x")
plt.grid(True)
plt.legend()
plt.show()
```

## Problem 9

Use a computer to plot the four polynomials 1, $x$, and the two you found in parts (b) $\big(f(x)=3x^2-1\big)$ and (c) $\big(f(x)=x^3-\frac{3}{5}x\big)$ on the same graph over $[−1, 1]$.

```python
x = np.linspace(-1,1,100)

plt.axhline(y=1, linestyle='--', label='f(x)=1')
plt.plot(x, x, "k--", label="f(x) = x")
plt.plot(x, 3*x**2 -1, "k--", label="f(x) = 3x^2-1")
plt.plot(x, x**3 - (3/5)*x, "k--", label="f(x) = x^3 - 3/5 x")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Comparison of orthogonal functions")
plt.grid(True)
plt.legend()
plt.show()
```
